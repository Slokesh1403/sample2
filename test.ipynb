```python
inputdata = ('1', '2', '3')
outputdata = tuple(map(lambda x: str(int(x) + 1), inputdata))
print(outputdata)

```

```python

inputdata = ("1", "2", "3")
outputdata = tuple(str(int(x) + 1) for x in inputdata)
print(outputdata)

```

```python
x = ("1", "2", "3")
result = tuple([int(i) + 1 for i in x])
print(result)
```

```python

a = ("1", "2", "3")
b = []
for x in a:
    b.append(str(int(x) + 1))
b = tuple(b)
print(a, "=", b)
```
